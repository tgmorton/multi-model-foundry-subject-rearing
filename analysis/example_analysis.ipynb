## Surprisal Analysis

Compare surprisal values across experimental conditions.

```python
# Example surprisal comparison
conditions = ["baseline", "no_expletives", "no_articles", "no_subject_pronouns"]
example_surprisals = [4.2, 4.5, 4.8, 5.1]  # Placeholder values

fig, ax = plt.subplots(figsize=(10, 6))
ax.bar(conditions, example_surprisals, color="steelblue")
ax.set_xlabel("Experimental Condition")
ax.set_ylabel("Mean Surprisal (bits)")
ax.set_title("Surprisal by Training Condition")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## BLIMP Performance

Analyze grammaticality judgment accuracy across conditions.

```python
# Example BLIMP accuracy data
blimp_categories = ["anaphor_agreement", "argument_structure", "binding", 
                    "control_raising", "determiner_noun_agreement"]
baseline_acc = [0.72, 0.68, 0.65, 0.71, 0.78]
no_expletives_acc = [0.70, 0.64, 0.63, 0.68, 0.75]

x = np.arange(len(blimp_categories))
width = 0.35

fig, ax = plt.subplots(figsize=(12, 6))
ax.bar(x - width/2, baseline_acc, width, label="Baseline", color="steelblue")
ax.bar(x + width/2, no_expletives_acc, width, label="No Expletives", color="coral")
ax.set_xlabel("BLIMP Category")
ax.set_ylabel("Accuracy")
ax.set_title("BLIMP Accuracy by Condition")
ax.set_xticks(x)
ax.set_xticklabels(blimp_categories, rotation=45, ha="right")
ax.legend()
ax.axhline(y=0.5, color="gray", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
